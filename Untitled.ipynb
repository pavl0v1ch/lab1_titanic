```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# У оригинального файла с UCI нет заголовков — задаём их вручную
column_names = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 
                'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']

df = pd.read_csv("heart.csv", names=column_names)

# В этом датасете пропуски обозначены символом '?' — заменяем на стандартные NaN
df = df.replace('?', np.nan)

display(df.head())
print("\nИнформация о датасете:")
df.info()
```

```python
print("Количество пропущенных значений ДО обработки:")
print(df.isnull().sum())
print(f"\nВсего пропусков в датасете: {df.isnull().sum().sum()}")
print(f"Пропуски в столбце 'ca': {df['ca'].isnull().sum()}")
print(f"Пропуски в столбце 'thal': {df['thal'].isnull().sum()}")
```

```python
# Целевая переменная для классификации (наличие болезни)
y_class = df['target']

# Целевая переменная для регрессии (например, уровень холестерина 'chol')
y_reg = df['chol']

# Признаки (удаляем обе целевые переменные)
X = df.drop(['target', 'chol'], axis=1)

# РАЗДЕЛЯЕМ ДАННЫЕ СРАЗУ! До заполнения пропусков и нормализации — защита от утечки данных
# stratify=y_class сохраняет пропорции классов (важно для ROC-кривой в 3 лабе)
X_train, X_test, y_class_train, y_class_test, y_reg_train, y_reg_test = train_test_split(
    X, y_class, y_reg, 
    test_size=0.2,       
    random_state=42,     
    stratify=y_class     
)

print(f"Размер обучающей выборки: {X_train.shape[0]} строк")
print(f"Размер тестовой выборки: {X_test.shape[0]} строк")

print("\nПропуски в X_train:")
print(X_train.isnull().sum())
print("\nПропуски в X_test:")
print(X_test.isnull().sum())
```

```python
print("=== ЗАПОЛНЕНИЕ ПРОПУСКОВ ===\n")

# 1. Числовые признаки — заполняем МЕДИАНОЙ, вычисленной ТОЛЬКО по X_train
num_cols = ['age', 'trestbps', 'thalach', 'oldpeak']
for col in num_cols:
    median_val = X_train[col].median()
    print(f"Столбец '{col}': медиана по train = {median_val:.2f}")
    X_train[col] = X_train[col].fillna(median_val)
    X_test[col] = X_test[col].fillna(median_val)

# 2. Категориальные признаки с пропусками ('ca', 'thal') — заполняем МОДОЙ из X_train
cat_cols_with_nan = ['ca', 'thal']
for col in cat_cols_with_nan:
    mode_val = X_train[col].mode()[0]  # mode() возвращает Series, берём [0]
    print(f"Столбец '{col}': мода по train = {mode_val}")
    X_train[col] = X_train[col].fillna(mode_val)
    X_test[col] = X_test[col].fillna(mode_val)

# Проверяем, что пропуски действительно заполнены (пункт 4 задания)
print("\n=== ПРОВЕРКА ПОСЛЕ ЗАПОЛНЕНИЯ ===")
print(f"Пропуски в X_train: {X_train.isnull().sum().sum()}")
print(f"Пропуски в X_test:  {X_test.isnull().sum().sum()}")
```

```python
# Все категориальные столбцы
all_cat_cols = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']

# drop_first=True убирает мультиколлинеарность (нужно для Линейной регрессии во 2 лабе)
# dtype=int делает столбцы 0/1 вместо True/False
X_train_encoded = pd.get_dummies(X_train, columns=all_cat_cols, drop_first=True, dtype=int)
X_test_encoded = pd.get_dummies(X_test, columns=all_cat_cols, drop_first=True, dtype=int)

# Выравниваем столбцы (если в test встретилась редкая категория, которой не было в train)
X_train_encoded, X_test_encoded = X_train_encoded.align(X_test_encoded, join='left', axis=1, fill_value=0)

print(f"Столбцов после OHE: {X_train_encoded.shape[1]}")
print("Названия столбцов:", X_train_encoded.columns.tolist())
```

```python
scaler = StandardScaler()

# Обучаем скалер ТОЛЬКО на train и сразу преобразуем
X_train_scaled = scaler.fit_transform(X_train_encoded)

# Преобразуем test, используя параметры, выученные на train (не подглядываем в тест!)
X_test_scaled = scaler.transform(X_test_encoded)

# Возвращаем формат DataFrame для удобства
X_train_final = pd.DataFrame(X_train_scaled, columns=X_train_encoded.columns)
X_test_final = pd.DataFrame(X_test_scaled, columns=X_test_encoded.columns)

display(X_train_final.head())
```

```python
# Сохраняем всё в отдельные CSV файлы
# Во 2 и 3 лабах ты просто сделаешь pd.read_csv() и сразу начнёшь обучать модели!

X_train_final.to_csv('X_train_processed.csv', index=False)
X_test_final.to_csv('X_test_processed.csv', index=False)

y_class_train.to_csv('y_class_train.csv', index=False)
y_class_test.to_csv('y_class_test.csv', index=False)

y_reg_train.to_csv('y_reg_train.csv', index=False)
y_reg_test.to_csv('y_reg_test.csv', index=False)

print("✅ Все файлы успешно сохранены в папку проекта!")
print("Готово к Лабораторным 2 и 3.")
```
